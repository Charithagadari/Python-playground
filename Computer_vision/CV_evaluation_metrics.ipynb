```python
def iou(box1, box2):
    x1,y1,x2,y2= box1
    x3,y3,x4,y4 = box2
    x_i1 = max(x1,x3)
    x_i2 = min (x2,x4)
    y_i1 = max (y1,y3)
    y_i2 = min(y2,y4)
    width = max(0,x_i2-x_i1)
    height = max (0, y_i2-y_i1)
    area_i = width * height
    area_u = (x2-x1) * (y2-y1) + (x4-x3) * (y4-y3) - area_i
    iou = area_i/area_u
    return(iou)
    
box1=1,4,6,8
box2 = 1,3,6,6
iou([1,4,6,8],[1,3,6,6])    
```

```python
def nms(scores, iou_threshold=0.5):

    keep = []

    sorted_scores = sorted(
        scores.items(),
        key=lambda x: x[1],
        reverse=True
    )

    while sorted_scores:

        current_box, current_score = sorted_scores.pop(0)

        keep.append((current_box, current_score))

        remaining = []

        for box, score in sorted_scores:

            if iou(current_box, box) < iou_threshold:
                remaining.append((box, score))

        sorted_scores = remaining

    return keep
```

```python
scores = {
    (1, 3, 4, 5): 0.45,
    (3, 5, 6, 8): 0.32,
    (10, 10, 20, 20): 0.95
}

print(nms(scores))
```
